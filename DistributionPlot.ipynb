```python

import seaborn as sns
import matplotlib.pyplot as plt
```

We use distribution plots to visualize the distribution of quantitative data.

Let's discuss some plots that allow us to visualize the distribution of a data set. These plots are:

distplot/histplot
jointplot
pairplot
rugplot

```python
df = sns.load_dataset('tips')
df
```

```python
df['size'].unique()
# this is quantative as unique values are very less
```

```python
sns.histplot(df['total_bill'],bins = 20,kde = True)
```

```python
plt.subplot(1,2,1)
sns.histplot(df['total_bill'],bins = 20,kde = True)
plt.subplot(1,2,2)
sns.histplot(df['tip'],bins = 20,kde = True)
```

JOINTPLOT --> ek sath 2 columns

```python
sns.jointplot(x = 'total_bill',y = 'tip',data = df)
sns.jointplot(x = 'total_bill',y = 'tip',data = df,kind = 'hex')
sns.jointplot(x = 'total_bill',y = 'tip',data = df,kind = 'reg')
sns.jointplot(x = 'total_bill',y = 'tip',data = df,kind = 'resid')
```

```python
# ye numerical columns me pair bnake graphs banata hai
# hue --> this will divide the graph on basis of categorical data
sns.pairplot(df,hue = 'size',palette='rainbow')
```
