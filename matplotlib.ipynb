```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
```

```python
x = np.linspace(0,5,11)
x
y = x**2
y
```

```python
plt.plot(x,y)
plt.title("hello")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
```

```python
# ------------------------           subplots            ---------------------------------
```

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,4)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,2)
```

access the canvas    

```python
fig = plt.figure(figsize=(12,8),dpi = 100)
axis1 = fig.add_axes([0.1,0.1,1,1])
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.1,0.9,0.8])
# clock wise from left side 
axis2.plot(x,y)
```

types of plots

```python
plt.scatter(x,y)
```

```python
plt.hist(x)
```

```python
plt.boxplot(x)
```

               saving plots --> saved as png inside working directory

```python
fig = plt.figure()
ax = fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)
plt.savefig("basicplot.png")
```
